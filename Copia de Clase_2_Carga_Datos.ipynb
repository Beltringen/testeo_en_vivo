## 🧠 Conceptos claves que estamos usando

Antes de seguir avanzando, es importante entender qué está pasando detrás del código que acabamos de ejecutar.

No necesitas memorizar todo, solo comprender la lógica general 👇

### 📦 Librerías / Paquetes


Las librerías son **conjuntos de herramientas ya creadas** que nos ayudan a trabajar más rápido.

En este caso usamos:

```python
import pandas as pd
```

👉 `pandas` es una librería especializada en análisis de datos
👉 Nos permite trabajar con tablas, limpiar datos, analizarlos, etc.

💡 En lugar de crear todo desde cero, usamos herramientas que ya existen.

### 📊 Objetos (DataFrames)

Cuando cargamos los datos:

```python
pacientes = pd.read_csv("pacientes.csv")
```

Estamos creando un **objeto**, específicamente un **DataFrame**.

👉 Un DataFrame es como una tabla de Excel dentro de Python
👉 Tiene filas, columnas y datos estructurados

En este caso:

* `pacientes` → es una tabla con información de pacientes
* `centros` → es otra tabla con información de hospitales

### ⚙️ Métodos

Los métodos son **acciones que podemos aplicar sobre un objeto**.

Ejemplo:

```python
pacientes.head()
```

👉 `head()` es un método
👉 Nos muestra las primeras filas de la tabla

💡 Es como decir:
“objeto → haz esto”

### 🔗 Estructura clave (MUY importante)

```python
objeto.metodo()
```

Ejemplo real:

```python
pacientes.head()
```

* `pacientes` → objeto
* `head()` → método

👉 Esta estructura la usarás TODO el tiempo en Python


### 🔍 Parámetros

Algunos métodos pueden recibir información adicional:

```python
pacientes.head(10)
```

👉 Aquí le decimos:
“muéstrame 10 filas en lugar de 5”

💡 A eso se le llaman **parámetros o argumentos**

### 📈 Interpretación (lo más importante)

Más allá del código, lo importante es entender qué estamos viendo:

* Cada fila = una atención de un paciente
* Cada columna = una característica (edad, enfermedad, etc.)
* Hay datos faltantes (NaN)
* Hay posibles inconsistencias

👉 Esto es exactamente lo que recibe un analista en la vida real
